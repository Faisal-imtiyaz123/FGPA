```python
arr = [["Hydrometeorological Forecasting	Free Elective",2.00,5],
       ["Advanced Computing in Climate Change	Free Elective",1.00,7],
       ["High Performance Computing in Weather Research and Forecast/GCMs	Free Elective",1.00,8],
    #    ["Introduction to HDL	Departmental Elective",1.00,""],
       ["Information Theory	Departmental Elective",3.00,5],
       ["FPGA Lab	Departmental Elective",2.00,7],
       ["Marketing Research	Free Elective",1.00,5],
       ["Supply Chain Management & Circular Economy	Free Elective",1.00,6],
       ["Elementary Linear Algebra	Free Elective",1.00,5],
       ["Predictive Modelling	Free Elective",3.00,7],
       ["Power System Analysis",3.00,5]]

cgpa = 0
total_credits = 0
for course in arr:
    total_credits+=course[1]
    if course[2]!="":
        cgpa+=course[2]*course[1]
cgpa = cgpa/total_credits
cumulative_cgpa = (7.29*70+cgpa*total_credits)/(70+total_credits)


print(f"cumulative cgpa: {cumulative_cgpa}, total credits : {total_credits}")
test_cgpa = 5.88
test_cumulative_cgpa = (7.287*70+test_cgpa*total_credits)/(70+total_credits)
print(f"test cumulative cgpa: {test_cumulative_cgpa}")
```
