```python
import numpy as np

def fft8(x):
    """Simple 8-point Radix-2 FFT"""
    # Bit-reversal permutation
    x = x[[0, 4, 2, 6, 1, 5, 3, 7]]
    
    # Stage 1: length-2 butterflies
    for i in range(0, 8, 2):
        t = x[i+1]
        x[i+1] = x[i] - t
        x[i] = x[i] + t
    
    # Stage 2: length-4 butterflies
    for i in range(0, 8, 4):
        t = x[i+2] * np.exp(-2j * np.pi * 0 / 4)
        x[i+2] = x[i] - t
        x[i] = x[i] + t
        
        t = x[i+3] * np.exp(-2j * np.pi * 1 / 4)
        x[i+3] = x[i+1] - t
        x[i+1] = x[i+1] + t
    
    # Stage 3: length-8 butterflies
    W = np.exp(-2j * np.pi * np.arange(8) / 8)
    t0 = x[4] * W[0]
    t1 = x[5] * W[1]
    t2 = x[6] * W[2]
    t3 = x[7] * W[3]
    
    x[4] = x[0] - t0
    x[0] = x[0] + t0
    x[5] = x[1] - t1
    x[1] = x[1] + t1
    x[6] = x[2] - t2
    x[2] = x[2] + t2
    x[7] = x[3] - t3
    x[3] = x[3] + t3
    
    return x

# Test
signal = np.array([1, 0, 0, 0, 0, 0, 0, 0])  # Impulse
result = fft8(signal)
print("FFT result:", result)
print("Matches NumPy:", np.allclose(result, np.fft.fft(signal)))
```
